## Parte 2. Análisis exploratorio


```python
# Primeras cinco filas
df.head()
```

```python
# Cantidad de filas y columnas
print('Filas:', df.shape[0])
print('Columnas:', df.shape[1])
```

```python
# Nombres de todas las columnas
list(df.columns)
```

```python
# Estructura del conjunto de datos y tipos de datos
df.info()
```

```python
# Estadisticas descriptivas de las columnas numericas
df.describe().round(2)
```

```python
# Valores nulos por columna
df.isnull().sum()
```

```python
# Registros duplicados
print('Filas duplicadas:', df.duplicated().sum())
```

```python
# Conteo por categoria relevante para el proyecto
print('Solicitudes por tipo de producto:')
print(df['Tipo_Producto'].value_counts())
print()
print('Resultado de la decision:')
print(df['Resultado_Decision'].value_counts())
print()
print('Solicitudes por categoria de riesgo:')
print(df['Categoria_Riesgo'].value_counts())
```

```python
# Suma, promedio, minimo y maximo del Monto_Solicitado
m = df['Monto_Solicitado']
print('Suma total  :', round(m.sum(), 2))
print('Promedio    :', round(m.mean(), 2))
print('Minimo      :', round(m.min(), 2))
print('Maximo      :', round(m.max(), 2))
```

## Parte 3. Resumen de los principales resultados


```python
tasa = (df['Resultado_Decision'] != 'Rechazado').mean() * 100
print('Total de solicitudes :', len(df))
print('Monto total          : RD$', round(m.sum(), 2))
print('Tasa de aprobacion   : {:.2f} %'.format(tasa))
print('Riesgo promedio      :', round(df['Score_Interno'].mean(), 2))
print('Producto top         :', df['Tipo_Producto'].value_counts().index[0])
```
