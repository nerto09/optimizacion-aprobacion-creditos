```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv(list(subido.keys())[0])
df.head()
```

### 1. Gráfico de barras — solicitudes por tipo de producto


```python
vc = df['Tipo_Producto'].value_counts()
plt.figure(figsize=(8,5))
plt.bar(vc.index, vc.values, color='#1B3A6B')
plt.title('Solicitudes por tipo de producto')
plt.ylabel('Numero de solicitudes'); plt.xticks(rotation=20, ha='right')
plt.tight_layout(); plt.show()
```

### 2. Gráfico de líneas — evolución mensual


```python
df['Fecha_Solicitud'] = pd.to_datetime(df['Fecha_Solicitud'])
serie = df.set_index('Fecha_Solicitud').resample('MS').size()
plt.figure(figsize=(8,5))
plt.plot(serie.index, serie.values, marker='o', color='#1B3A6B')
plt.title('Evolucion mensual de las solicitudes')
plt.ylabel('Numero de solicitudes'); plt.tight_layout(); plt.show()
```

### 3. Histograma — distribución del monto solicitado


```python
plt.figure(figsize=(8,5))
plt.hist(df['Monto_Solicitado']/1e6, bins=12, color='#2E5A94', edgecolor='white')
plt.title('Distribucion del monto solicitado')
plt.xlabel('Monto (millones de RD$)'); plt.ylabel('Solicitudes')
plt.tight_layout(); plt.show()
```

### 4. Gráfico de dona — resultado de la decisión


```python
rc = df['Resultado_Decision'].value_counts()
plt.figure(figsize=(7,6))
plt.pie(rc.values, labels=rc.index, autopct='%1.0f%%',
        startangle=90, wedgeprops=dict(width=0.42))
plt.title('Distribucion de los resultados de la decision'); plt.show()
```

### 5. Gráfico de dispersión — riesgo vs monto


```python
plt.figure(figsize=(8.5,5.5))
for res in df['Resultado_Decision'].unique():
    s = df[df['Resultado_Decision'] == res]
    plt.scatter(s['Score_Interno'], s['Monto_Solicitado']/1e6, label=res, s=70)
plt.xlabel('Score de riesgo interno'); plt.ylabel('Monto (millones de RD$)')
plt.legend(title='Resultado'); plt.title('Riesgo interno vs monto solicitado')
plt.tight_layout(); plt.show()
```
